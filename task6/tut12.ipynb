```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from datetime import datetime
from zoneinfo import ZoneInfo
```

```python
# Step 1: Load the Dataset
apps_df = pd.read_csv(r"C:\Users\DELL\Pictures\elevance_skills\play_store_data\apps.csv")
reviews_df = pd.read_csv(r"C:\Users\DELL\Pictures\elevance_skills\play_store_data\user_reviews.csv")
```

```python
apps_df.head()
```

```python
bar_data = apps_df.copy()

bar_data["Last Updated"] = pd.to_datetime(
    bar_data["Last Updated"],
    errors="coerce"
)

bar_data = bar_data.dropna(
    subset=["Last Updated"]
)
```

```python
bar_data = bar_data[
    bar_data["Last Updated"].dt.month == 1
]
```

```python
bar_data["Size"] = (
    bar_data["Size"]
    .astype(str)
    .str.replace("M", "", regex=False)
    .str.replace("k", "", regex=False)
    .str.strip()
)

bar_data["Size"] = pd.to_numeric(
    bar_data["Size"],
    errors="coerce"
)
```

```python
bar_data = bar_data.dropna(
    subset=["Size"]
)

bar_data = bar_data[
    bar_data["Size"] < 10
]
```

```python
bar_data["Rating"] = pd.to_numeric(
    bar_data["Rating"],
    errors="coerce"
)


bar_data["Reviews"] = pd.to_numeric(
    bar_data["Reviews"],
    errors="coerce"
)


bar_data["Installs"] = (
    bar_data["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)

bar_data["Installs"] = pd.to_numeric(
    bar_data["Installs"],
    errors="coerce"
)


bar_data = bar_data.dropna(
    subset=[
        "Rating",
        "Reviews",
        "Installs"
    ]
)
```

```python
category_data = (
    bar_data
    .groupby("Category")
    .agg(
        Average_Rating=(
            "Rating",
            "mean"
        ),
        
        Total_Reviews=(
            "Reviews",
            "sum"
        ),
        
        Total_Installs=(
            "Installs",
            "sum"
        )
    )
    .reset_index()
)
```

```python
category_data = category_data[
    category_data["Average_Rating"] >= 4.0
]

category_data
```

```python
top_categories = (
    category_data
    .sort_values(
        "Total_Installs",
        ascending=False
    )
    .head(10)
)
```

```python
top_categories
```

```python
top_categories = top_categories.copy()

top_categories["Total_Reviews_Millions"] = (
    top_categories["Total_Reviews"]
    / 1000000
)


top_categories = top_categories.sort_values(
    "Total_Installs",
    ascending=False
)


top_categories
```

```python
# Get current time in IST

india_time = datetime.now(
    ZoneInfo("Asia/Kolkata")
)

current_hour = india_time.hour


# Show graph only between 3 PM and 5 PM IST

if 15 <= current_hour < 17:

    
    # Create positions for categories
    
    x = np.arange(
        len(top_categories)
    )
    
    
    # Width of bars
    
    width = 0.4
    
    
    # Create figure
    
    plt.figure(
        figsize=(14, 7)
    )
    
    
    # Average Rating bars
    
    plt.bar(
        x - width / 2,
        top_categories[
            "Average_Rating"
        ],
        width,
        label="Average Rating"
    )
    
    
    # Total Reviews bars
    
    plt.bar(
        x + width / 2,
        top_categories[
            "Total_Reviews_Millions"
        ],
        width,
        label="Total Reviews (Millions)"
    )
    
    
    # Category names
    
    plt.xticks(
        x,
        top_categories["Category"],
        rotation=45,
        ha="right"
    )
    
    
    # Title
    
    plt.title(
        "Average Rating and Total Reviews for Top 10 App Categories by Installs"
    )
    
    
    # X-axis
    
    plt.xlabel(
        "App Category"
    )
    
    
    # Y-axis
    
    plt.ylabel(
        "Value"
    )
    
    
    # Legend
    
    plt.legend()
    
    
    plt.tight_layout()
    
    
    # Display graph below Jupyter cell
    
    plt.show()


else:

    print(
        "This graph is available only between "
        "3 PM IST and 5 PM IST."
    )
```
