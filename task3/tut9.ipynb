```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
apps_df = pd.read_csv(r"C:\Users\DELL\Pictures\elevance_skills\play_store_data\apps.csv")

apps_df.head()
```

```python
apps_df.columns
```

```python
apps_df = apps_df.drop(columns=['Unnamed: 0'])
```

```python
apps_df = apps_df.dropna(
    subset=['Category', 'Installs']
)
```

```python
apps_df['Installs'] = (
    apps_df['Installs']
    .str.replace(',', '')
    .str.replace('+', '')
)

apps_df['Installs'] = apps_df['Installs'].astype(int)
```

```python
apps_df[['Category', 'Installs']].head()
```

```python
filtered_df = apps_df[
    ~apps_df['Category'].str.startswith(
        ('A', 'C', 'G', 'S')
    )
]
```

```python
category_installs = (
    filtered_df
    .groupby('Category')['Installs']
    .sum()
    .reset_index()
)
```

```python
top_5_categories = (
    category_installs
    .sort_values(
        by='Installs',
        ascending=False
    )
    .head(5)
)
```

```python
top_5_categories = (
    category_installs
    .sort_values(
        by='Installs',
        ascending=False
    )
    .head(5)
)
```

```python
top_5_categories
```

```python
top_5_categories['Highlight'] = np.where(
    top_5_categories['Installs'] > 1000000,
    'Above 1 Million',
    'Below 1 Million'
)
```

```python
top_5_categories
```

```python
fig = px.bar(
    top_5_categories,
    x='Category',
    y='Installs',
    color='Highlight',
    hover_data=['Installs'],
    title='Top 5 App Categories by Installs'
)

fig.show()
```
